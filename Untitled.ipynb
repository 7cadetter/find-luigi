```python
%matplotlib inline

import cv2
import numpy as np
from matplotlib import pyplot as plt
from matplotlib import image as image

def show(I):
    cv2.imshow("image", I)
    key = cv2.waitKey(0)
    cv2.destroyAllWindows()
    cv2.waitKey(1)
```

```python
def binarize(gray, invert=False):
    blur = cv2.GaussianBlur(gray, (5, 5), 0)
    mode = cv2.THRESH_BINARY_INV if invert else cv2.THRESH_BINARY
    _, th = cv2.threshold(blur, 0, 255, mode + cv2.THRESH_OTSU)
    return th

def template_contour(path):
    img = cv2.imread(path, cv2.IMREAD_UNCHANGED)
    if img.ndim == 3 and img.shape[2] == 4:
        # use transparency as the shape mask
        _, binary = cv2.threshold(img[:, :, 3], 127, 255, cv2.THRESH_BINARY)
    else:
        gray = img if img.ndim == 2 else cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
        binary = binarize(gray)
        # make sure the shape is the minority (white) region
        if np.count_nonzero(binary) > binary.size / 2:
            binary = cv2.bitwise_not(binary)
    contours, _ = cv2.findContours(binary, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_NONE)
    return max(contours, key=cv2.contourArea)

def find_matches(scene_bgr, template_cnt, max_score=0.1, min_area=100):
    gray = cv2.cvtColor(scene_bgr, cv2.COLOR_BGR2GRAY)
    results = []
    for invert in (False, True):          # try dark-on-light and light-on-dark
        th = binarize(gray, invert)
        contours, _ = cv2.findContours(th, cv2.RETR_LIST, cv2.CHAIN_APPROX_NONE)
        for c in contours:
            if cv2.contourArea(c) < min_area:
                continue
            score = cv2.matchShapes(template_cnt, c, cv2.CONTOURS_MATCH_I1, 0.0)
            results.append((score, c))
    results.sort(key=lambda r: r[0])
    return [r for r in results if r[0] <= max_score], results

template = template_contour('img/icon.png')
#scene = cv2.imread('img/i-squaresmall.png')


Video = cv2.VideoCapture('vid/yt_clip.mp4')
(check, scene) = Video.read()
matches, all_results = find_matches(scene, template)

while check:

    # See what scores you're actually getting, useful for tuning max_score
    print("Best 5 scores:", [round(s, 4) for s, _ in all_results[:5]])
    
    out = scene.copy()
    for score, c in matches:
        x, y, w, h = cv2.boundingRect(c)
        print(f"Match: score={score:.4f} at ({x},{y}) size {w}x{h}")
        cv2.rectangle(out, (x, y), (x + w, y + h), (0, 255, 0), 2)
    
    plt.imshow(cv2.cvtColor(out, cv2.COLOR_BGR2RGB))
    plt.axis('off')
    plt.show()


    key = cv2.waitKey(1)

    # if the 'q' keqy is pressed, quit:
    if key == ord("q"):
         break

    # Next Frame:
    (check, I) = Video.read()

Video.release()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def find_blobs(img, thresh=30, min_area=500):
    """Segment bright objects on the black background."""
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    gray = cv2.GaussianBlur(gray, (5, 5), 0)
    _, th = cv2.threshold(gray, thresh, 255, cv2.THRESH_BINARY)
    th = cv2.morphologyEx(th, cv2.MORPH_CLOSE, np.ones((5, 5), np.uint8))
    contours, _ = cv2.findContours(th, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_NONE)
    return [c for c in contours if cv2.contourArea(c) >= min_area], th

def masked_crop(img, cnt):
    """Cut out the object's bounding box, blacking out everything outside its contour."""
    mask = np.zeros(img.shape[:2], np.uint8)
    cv2.drawContours(mask, [cnt], -1, 255, cv2.FILLED)
    x, y, w, h = cv2.boundingRect(cnt)
    return cv2.bitwise_and(img, img, mask=mask)[y:y+h, x:x+w]


def ORB(icon, sample):
    
    
    
    icon_bw = cv2.cvtColor(icon, cv2.COLOR_BGR2GRAY)
    sample_bw = cv2.cvtColor(sample, cv2.COLOR_BGR2GRAY)
    
    # Gaussian blur (noise reduction)
    icon_bl = cv2.GaussianBlur(icon_bw, (7, 7), 0)
    sample_bl = cv2.GaussianBlur(sample_bw, (7, 7), 0)
    
    show(icon_bw)
    show(icon_bl)


    # FILTER FOR GREEN - USED
    hsv = cv2.cvtColor(sample, cv2.COLOR_BGR2HSV)  # no blur, so tiny bits of cap survive
    green = cv2.inRange(hsv, np.array([40, 80, 60]), np.array([85, 255, 255]))
    
    # Dilation: grow each green pixel into a face-sized region so the whole face is searched
    FACE_WIDTH = 50  # width of one face in the scene, in pixels
    face_kernel = cv2.getStructuringElement(cv2.MORPH_ELLIPSE, (2 * FACE_WIDTH, 2 * FACE_WIDTH))
    
    # Shape analysis: remove large, round green blobs (Yoshi heads); keep everything else
    cnts, _ = cv2.findContours(green, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_NONE)
    for c in cnts:
        area = cv2.contourArea(c)
        perimeter = cv2.arcLength(c, True)
        circularity = 4 * np.pi * area / perimeter**2 if perimeter else 0  # 1.0 = perfect circle
        if area > 0.4 * FACE_WIDTH**2 and circularity > 0.7:
            cv2.drawContours(green, [c], -1, 0, thickness=cv2.FILLED)
            
    orb_mask = cv2.dilate(green, face_kernel)
    
    orb = cv2.ORB_create(nfeatures=10000)
    
    queryKeypoints, queryDescriptors = orb.detectAndCompute(icon_bw, None)
    trainKeypoints, trainDescriptors = orb.detectAndCompute(sample_bw, orb_mask)
    
    matcher = cv2.BFMatcher(cv2.NORM_HAMMING)
    matches = matcher.match(queryDescriptors, trainDescriptors)
    matches = sorted(matches, key=lambda x: x.distance)
    
    best_matches = matches[:50]
    
    final_img = cv2.drawMatches(
        icon, queryKeypoints,
        sample, trainKeypoints,
        best_matches,
        None
    )
    
    final_img = cv2.resize(final_img, (1000, 650))
    
    print(icon.shape)
    print(icon_bw.min(), icon_bw.max())
    print("Query keypoints:", len(queryKeypoints))
    print("Train keypoints:", len(trainKeypoints))
    print("Matches:", len(matches))
    
    print([m.distance for m in matches[:10]])
    
    src_pts = np.float32([
        queryKeypoints[m.queryIdx].pt for m in best_matches
    ]).reshape(-1, 1, 2)
    
    dst_pts = np.float32([
        trainKeypoints[m.trainIdx].pt for m in best_matches
    ]).reshape(-1, 1, 2)
    
    H, mask = cv2.findHomography(src_pts, dst_pts, cv2.RANSAC, 5.0)
    
    h, w = icon.shape[:2]
    
    corners = np.float32([
        [0, 0],
        [w, 0],
        [w, h],
        [0, h]
    ]).reshape(-1, 1, 2)
    
    transformed_corners = cv2.perspectiveTransform(corners, H)
    
    result = sample.copy()
    
    pts = transformed_corners.reshape(-1, 2)
    cx, cy = pts.mean(axis=0)
    
    sides = [
        np.linalg.norm(pts[i] - pts[(i + 1) % 4])
        for i in range(4)
    ]
    
    r = int(np.clip(np.mean(sides) / 2 * 2.0, 100, 125))
    
    cv2.circle(result, (int(cx), int(cy)), r, (0, 0, 255), 3)
    plt.figure(figsize=(12, 8))
    plt.imshow(cv2.cvtColor(result, cv2.COLOR_BGR2RGB))
    plt.show()

template_img = cv2.imread('img/icon.png')
scene = cv2.imread('img/covered_1.png')


#Video = cv2.VideoCapture('vid/yt_clip.mp4')
#(check, scene) = Video.read()
#matches, all_results = find_matches(scene, template)

# Template: largest blob in the icon
t_blobs, _ = find_blobs(template_img)
t_cnt = max(t_blobs, key=cv2.contourArea)
t_crop = masked_crop(template_img, t_cnt)
th_, tw_ = t_crop.shape[:2]
    
# Scene: every blob is a candidate
blobs, binary = find_blobs(scene)
    
results = []
for c in blobs:
    crop = masked_crop(scene, c)
    crop = cv2.resize(crop, (tw_, th_), interpolation=cv2.INTER_AREA)
    
    # colour/detail similarity (1.0 = identical); same-size inputs give a single value
    corr = cv2.matchTemplate(crop, t_crop, cv2.TM_CCOEFF_NORMED)[0][0]
    # shape-only score (lower = more similar), for comparison
    shape = cv2.matchShapes(t_cnt, c, cv2.CONTOURS_MATCH_I1, 0.0)
    
    results.append((corr, shape, c))
    
results.sort(key=lambda r: r[0], reverse=True)
    
#for corr, shape, c in results:
#    x, y, w, h = cv2.boundingRect(c)
#    print(f"at ({x},{y}) size {w}x{h}  correlation={corr:.3f}  shape_score={shape:.4f}")

best_corr, _, best_cnt = results[0]

if best_corr > 0.7: #if luigi is found with contours
    x, y, w, h = cv2.boundingRect(best_cnt)
    out = scene.copy()
    cv2.rectangle(out, (x, y), (x + w, y + h), (0, 0, 255), 10)
        
    fig, ax = plt.subplots(1, 3, figsize=(14, 5))
    ax[0].imshow(cv2.cvtColor(t_crop, cv2.COLOR_BGR2RGB)); 
    ax[0].set_title("template")
    ax[1].imshow(binary, cmap='gray');                      
    ax[1].set_title("segmentation")
    ax[2].imshow(cv2.cvtColor(out, cv2.COLOR_BGR2RGB));     
    ax[2].set_title(f"best match ({best_corr:.2f})")
    for a in ax: a.axis('off')
    plt.show()
else: #ORB
    ORB(template_img, scene)
    
```
